# Create a 1D NumPy Array

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
print(temperatures_w1)
```

```python
print("Shape: ", temperatures_w1.shape)
```

```python
print("Data type: ",temperatures_w1.dtype)
```

```python
print("No of elements: ",temperatures_w1.size)
```

# Array Operations

```python
temp_f = (temperatures_w1 * 9/5)+32
```

```python
print("Temperature in farenheit: ",temp_f)
```

```python
print("Max temperature: ",temperatures_w1.max())
```

```python
print("Min temperature: ",temperatures_w1.min())
```

```python
print("Mean temperature for the week: ",round(temperatures_w1.mean(),2))
```

```python
print("Temperature for first 3 days of the week: ",temperatures_w1[:3])
```

```python
print("Temperatures for the weekend (last two days): ",temperatures_w1[-2:])
```

# Pandas Series

```python
import pandas as pd
```

# Creating Pandas Series

```python
Marks = pd.Series([95, 92, 89, 85, 80],index =['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
```

```python
print(Marks)
```

# Indexing and Slicing

```python
print("Mark of the 1st rank student using the integer index position: ",Marks.iloc[0])
```

```python
print("Marks of the top 3 ranks by specifying their index labels: ",Marks.loc[['Rank1','Rank2','Rank3']])
```

```python
print("Mark of the 3rd rank student using iloc: ",Marks.iloc[2])
```

```python
print("Ranks where the marks are greater than 90: ",Marks[Marks>90])
```

# Manipulating Series

```python
Marks.loc['Rank1']=100
```

```python
print(Marks)
```

```python
Marks.drop('Rank5')
```

```python
cgpa =Marks/10
```

```python
print("CGPA: ",cgpa)
```

# Pandas DataFrame

# Creating Pandas DataFrame

```python
transactions =pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 
                    'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
 'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
 'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
```

# Data Exploration

```python
print(transactions)
```

```python
print("Head: ",transactions.head())
```

```python
print("Tail: ",transactions.tail())
```

```python
print("Shape: ",transactions.shape)
```

```python
print("Columns: ",transactions.columns)
```

```python
print("Data type: ",transactions.dtypes)
```

```python
print("Product Category and Amount \n ")
print(transactions[['ProductCategory', 'Amount']])
```

```python
print("Last 3 columns \n")
print(transactions.iloc[:,-3:])
```

```python
f = transactions[(transactions['Region']=='North')&(transactions['Amount']>200)]
print(f)
```

```python
print("Value counts \n")
print(transactions['ProductCategory'].value_counts())
```

```python
print("Unique values in Region \n ")
print(transactions['Region'].unique())
```

```python
print("Mean amount for each region \n ")
print(transactions.groupby ('Region')['Amount'].mean())
```

# Manipulating the DataFrame

```python
transactions.loc[transactions['TransactionID']==102,
   'Amount']=165
print(transactions)
```

```python
transactions['Discount']=transactions['Amount']*0.1
print(transactions)
```

```python
transactions[transactions['TransactionID'] != 109]
```

```python
transactions = transactions.drop(columns='Discount')
print(transactions)
```
